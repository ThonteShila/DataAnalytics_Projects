```python
import nltk
```

```python
nltk.download("vader_lexicon")
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
```

```python
sia=SentimentIntensityAnalyzer()
```

```python
sia
```

```python
comments["comment_text"]
```

```python
sia.polarity_scores("MY FAN . attendance")['compound']
```

```python
sentiments=[]
for i in comments["comment_text"]:
    score=sia.polarity_scores(str(i))['compound']
    sentiments.append(score)
sentiments[0:100]
```

```python
comments["Polarity"]=sentiments
```

```python
comments
```

#do punctuation in title affect views,likes

```python
import string
```

```python
string.punctuation
```

```python
len([char for char in full_df['title'][0] if char in string.punctuation])
```

```python
def punct_count(text):
   return len([char for char in text if char in string.punctuation])
```

```python
sample=full_df[0:100]
```

```python
sample['pct_count']=sample['title'].apply(punct_count)
```

```python
sample['pct_count']
```

#punctuation count is 3 or 4 views are more

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='pct_count',y='views',data=sample)
plt.show()
```
