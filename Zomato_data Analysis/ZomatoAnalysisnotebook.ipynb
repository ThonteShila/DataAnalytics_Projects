```python
#count how many type each word repeated in the reviews
#convert series 2D in 1D list

total_review=list(review_tokens_clean)
```

```python

type(total_review)
```

```python
total_review
```

```python
total_review_1D=[]
for i in total_review:
    for word in i:
        total_review_1D.append(word)
```

```python
total_review_1D
```

```python
from nltk import FreqDist
```

```python
fd=FreqDist()
fd
```

```python
for word in total_review_1D:
    fd[word] = fd[word] + 1
fd[word]
```

```python
fd.most_common(20)
```

```python
fd.plot(20)
```

```python
df
```

```python

#bigram and trigram analysis 
#bigream means two words together
#trigram means three words together 
from  nltk import bigrams,trigrams,FreqDist
bigrams_list=list(bigrams(total_review_1D))
trigrams_list=list(trigrams(total_review_1D))
    
```

```python

fd_bi=FreqDist()
for bigram in bigrams_list:
    fd_bi[bigram]+=1
    
fd_bi.most_common(20)
```

```python
fd_bi.plot(20)
```

```python
fd_tri=FreqDist()
for i in trigrams_list:
    fd_tri[i]+=1
fd_tri.most_common(20)
```

```python
fd_tri.plot(20)
```
