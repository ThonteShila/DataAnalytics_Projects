```python
team=match_df["team1"].value_counts()+match_df["team2"].value_counts()
team
```

```python
t_df=pd.DataFrame(team).reset_index().rename(columns={"index":"team","count":"matches"})
t_df
```

```python
win_df=pd.DataFrame(match_df["winner"].value_counts()).reset_index()
win_df.columns=["team","wins"]
```

```python

win_df
```

```python
data=t_df.merge(win_df,on="team",how="inner")
data
```

```python
traces1=go.Bar(x=data["team"],y=data["matches"],name="Total matches")
traces2=go.Bar(x=data["team"],y=data["wins"],name="Total wins")
```

```python
d=[traces1,traces2]
```
