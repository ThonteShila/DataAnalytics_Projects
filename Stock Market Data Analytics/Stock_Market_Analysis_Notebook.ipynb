#Which is safest stock to invest

```python
#std=stock fluctuate around their mean
```

```python
all_data["Daily Return"]=all_data.groupby("Name")["close"].pct_change()
```

```python
risk=all_data.groupby("Name")["Daily Return"].std()
```

```python
risk
```

```python
safestocks=risk.idxmin()
safestocks
```

#Which Stock made highest money

```python
all_data["Daily Return"]
```

```python
all_data["Cumulative_Return"]=(1+all_data["Daily Return"]).groupby(all_data["Name"]).cumprod()
```

```python
best=all_data.groupby("Name")["Cumulative_Return"].last() 
```

```python

best.idxmax()
```

```python
plt.bar(best.index,best.values)
plt.title("final value return while investing 1 rs investment in each stock")
```

#1 million safe investment at amazon

#To find most crash days

```python
all_data[all_data["Name"]=="AAPL"].nsmallest(5,"Daily Return")
```

```python
crash_df=all_data.groupby("Name",group_keys=False).apply(lambda x:x.nsmallest(5,"Daily Return"))
```

```python
crash_df
```

```python
px.bar(crash_df,x="date",y="Daily Return",color="Name",title="Worst 5 days for each stock",labels={"date":"Date","Daily Return":"Daily Return"})
```

```python

crash_df["crashlabel"]=crash_df["Name"]+"-"+crash_df["date"].dt.strftime("%Y-%m-%d")
```

```python
px.bar(crash_df,x="crashlabel",y="Daily Return",color="Name",title="Worst 5 days for each stock",labels={"crashlabel":"Stock-Date","Daily Return":"Daily Return"})
```

#Moving avg Trend analysis


```python
all_data["close"]
```

```python
all_data["Ma20"]=all_data.groupby("Name")["close"].transform(lambda  x: x.rolling(20).mean())
```

```python
all_data["Ma10"]=all_data.groupby("Name")["close"].transform(lambda  x: x.rolling(10).mean())
```

```python
all_data["ma50"]=all_data.groupby("Name")["close"].transform(lambda  x: x.rolling(50).mean())
```

```python
all_data
```

```python
avg_stock=all_data[all_data["Name"]=="AAPL"]
```

```python
avg_stock
```

```python
px.line(avg_stock, x="date",y=["close","Ma20","ma50","Ma10"],title="AAPL Stock Price with Moving Averages",labels={"date":"Date","value":"Price"},color_discrete_map={"close":"blue","Ma20":"orange","ma50":"green","Ma10":"red"})
```

#Best Month to invest in each stock.

```python
monthly_return=all_data.groupby(["Name","month"])["Daily Return"].mean().reset_index()
monthly_return
```

```python
px.bar(monthly_return,x="month",y="Daily Return",barmode="group",color="Name",title="Average Monthly Return for each stock",labels={"month":"Month","daily return":"Average Daily Return"})
```
